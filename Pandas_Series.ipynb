```python
import pandas as pd

#Series是某一列
#直接构建
f1 = pd.Series([1,2,3,4,5])
f1
```

```python
#自定义索引
f2 = pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
f2
```

```python
#也可以直接从DF提取
df4 = pd.DataFrame([
    ['张三', 85, 92, 78],
    ['李四', 78, 88, 95],
    ['王五', 92, 96, 89]
], columns=['姓名', '语文', '数学', '英语'], index=['a', 'b', 'c'])

f3 = df4['语文']
f3
```
